n = int(input())
m = int(input())
print(n ** m)

**Алгоритм** — это сама идея (последовательность действий) возведения числа в степень:
1. Получить число n.
2. Получить степень m.
3. Умножить n само на себя m раз.
4. Вывести результат.

**Программа** — это конкретная реализация алгоритма на языке Python:

n = int(input())
m = int(input())
print(n ** m)
